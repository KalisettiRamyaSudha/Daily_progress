```python
"""
W3D2 - Multiple Plot Practice
Creates three separate charts, each suited to a different kind of question.
"""

import matplotlib.pyplot as plt
import numpy as np

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [12000, 13500, 11000, 15000, 17500, 16000]
products = ["Laptop", "Mouse", "Keyboard", "Monitor"]
market_share = [45, 20, 15, 20]
rng = np.random.default_rng(seed=3)
delivery_times = rng.normal(loc=4.5, scale=1.2, size=200)  # days

# --- Chart 1: Line chart -> good for showing a trend over time ---
plt.figure(figsize=(7, 4))
plt.plot(months, revenue, marker="o", color="#4c72b0")
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.savefig("multiple_plots_line_W3D2.png", dpi=150)
print("Saved multiple_plots_line_W3D2.png -> purpose: show how revenue changes over time")

# --- Chart 2: Bar chart -> good for comparing discrete categories ---
plt.figure(figsize=(7, 4))
plt.bar(products, market_share, color="#55a868")
plt.title("Market Share by Product")
plt.xlabel("Product")
plt.ylabel("Market Share (%)")
plt.tight_layout()
plt.savefig("multiple_plots_bar_W3D2.png", dpi=150)
print("Saved multiple_plots_bar_W3D2.png -> purpose: compare a value across categories")

# --- Chart 3: Histogram -> good for showing the distribution/spread of a dataset ---
plt.figure(figsize=(7, 4))
plt.hist(delivery_times, bins=20, color="#dd8452", edgecolor="black")
plt.title("Distribution of Delivery Times")
plt.xlabel("Delivery Time (days)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("multiple_plots_hist_W3D2.png", dpi=150)
print("Saved multiple_plots_hist_W3D2.png -> purpose: show the spread/shape of a single variable")
```
