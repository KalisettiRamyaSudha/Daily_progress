```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(seed=42)
data = rng.normal(loc=70, scale=10, size=500)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for ax, bins in zip(axes, [10, 20, 40]):
    ax.hist(data, bins=bins, color="#55a868", edgecolor="black", alpha=0.8)
    ax.set_title(f"bins={bins}")
    ax.set_xlabel("Value")
    ax.set_ylabel("Frequency")
    ax.grid(axis="y", linestyle="--", alpha=0.5)

fig.suptitle("Histogram of Sample Data at Different Bin Counts", fontsize=14, fontweight="bold")
fig.tight_layout()
plt.show()
```
