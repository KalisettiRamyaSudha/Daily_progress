```python
"""
W3D2 - Scatter Plot
Plots two numerical columns and observes the relationship between them.
"""

import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(seed=10)
hours_studied = rng.uniform(1, 10, size=40)
# exam_score roughly increases with hours_studied, plus some random noise
exam_score = 40 + hours_studied * 5.5 + rng.normal(0, 5, size=40)

plt.figure(figsize=(8, 5))
plt.scatter(hours_studied, exam_score, color="#c44e52", edgecolor="black", alpha=0.8)

plt.title("Hours Studied vs. Exam Score", fontsize=14, fontweight="bold")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()

plt.savefig("scatter_plot_W3D2.png", dpi=150)
print("Saved scatter_plot_W3D2.png")

correlation = np.corrcoef(hours_studied, exam_score)[0, 1]
print(f"Correlation between hours studied and exam score: {correlation:.2f}")
print("-> Positive correlation: more hours studied tends to mean a higher score.")
```
