```python
"""
W3D2 - Plot Customization
Adds legends, gridlines, markers, and figure-size changes to improve readability.
"""

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
product_a_sales = [220, 240, 210, 260, 300, 280]
product_b_sales = [180, 200, 230, 210, 250, 270]

# Larger figure size for better readability
plt.figure(figsize=(10, 6))

# Distinct markers + colors per line, so they're distinguishable even in grayscale
plt.plot(months, product_a_sales, marker="o", linestyle="-", color="#4c72b0",
         linewidth=2, markersize=8, label="Product A")
plt.plot(months, product_b_sales, marker="s", linestyle="--", color="#c44e52",
         linewidth=2, markersize=8, label="Product B")

plt.title("Product Sales Comparison (Customized)", fontsize=15, fontweight="bold")
plt.xlabel("Month", fontsize=12)
plt.ylabel("Units Sold", fontsize=12)

# Gridlines for easier value reading
plt.grid(True, linestyle="--", linewidth=0.7, alpha=0.6)

# Legend so viewers know which line is which
plt.legend(loc="upper left", fontsize=11, frameon=True)

plt.tight_layout()
plt.savefig("plot_customization_W3D2.png", dpi=150)
print("Saved plot_customization_W3D2.png")
```
