```python
"""
W3D1 - Bar Chart Program
"""

import matplotlib.pyplot as plt

categories = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headset"]
units_sold = [45, 120, 80, 35, 60]

plt.figure(figsize=(8, 5))
bars = plt.bar(categories, units_sold, color="#4c72b0")

plt.title("Units Sold by Product", fontsize=14, fontweight="bold")
plt.xlabel("Product")
plt.ylabel("Units Sold")
plt.grid(axis="y", linestyle="--", alpha=0.5)

# Label each bar with its value
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, height + 1, str(height),
             ha="center", va="bottom")

plt.tight_layout()
plt.show()
```
